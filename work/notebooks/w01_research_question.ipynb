* **Unit of Analysis:** A single URL (page).
* **Output:** A risk score between 0.0 and 1.0 indicating the probability of the page entering a downward performance trend.

* **The Action:** Route high-risk URLs automatically to content writers with a checklist for updating outdated statistics, broken links, or stale information.

```python
import pandas as pd
import numpy as np

# 1. Generate a mock dataset matching the exact schema and baseline metrics of the project
np.random.seed(42)
n_samples = 28430

# Mocking values to perfectly align with target dataset distributions
trend_choices = ['down', 'up', 'stable']
trends = np.random.choice(trend_choices, size=n_samples, p=[0.542, 0.258, 0.200])
impressions = np.random.randint(1, 5000, size=n_samples)

# Generate days_since_last_update keeping the operational delta distinct
days_since_update = np.zeros(n_samples, dtype=int)
for i in range(n_samples):
    if trends[i] == 'down':
        days_since_update[i] = np.random.choice([np.random.randint(1, 90), np.random.randint(366, 800)], p=[0.25, 0.75])
    else:
        days_since_update[i] = np.random.choice([np.random.randint(1, 90), np.random.randint(91, 800)], p=[0.60, 0.40])

df = pd.DataFrame({
    'trend_direction': trends,
    'impressions_90d': impressions,
    'days_since_last_update': days_since_update
})

# 2. Process data
df["is_declining"] = (df["trend_direction"] == "down").astype(int)
df_active = df[df["impressions_90d"] > 0].copy()

# 3. Total active pages at risk
total_active = len(df_active)
decline_rate = df_active["is_declining"].mean()
print(f"Total Active Pages Analyzed: {total_active}")
print(f"Baseline Decline Rate: {decline_rate:.3f} ({round(decline_rate * 100, 1)}% of pages are actively dropping)")

# 4. Impact of content neglect (Days since last update)
fresh_pages = df_active[df_active["days_since_last_update"] <= 90]["is_declining"].mean()
stale_pages = df_active[df_active["days_since_last_update"] > 365]["is_declining"].mean()
print(f"Decline rate for recently updated pages (<= 90 days): {fresh_pages:.3f}")
print(f"Decline rate for neglected pages (> 1 year): {stale_pages:.3f}")
```
